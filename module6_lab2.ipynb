```python
import sqlite3
import pandas as pd

con = sqlite3.connect("module6_lab2.db")
print("Connected to SQLite database.")
```

```python
con.execute("DROP TABLE IF EXISTS customers")

con.execute("""
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    total_spent REAL
)
""")

print("customers table created.")
```

```python
customers = [
    (1, "Alex Carter", "Los Angeles", 245.75),
    (2, "Maya Brooks", "Santa Monica", 389.20),
    (3, "Jordan Lee", "Inglewood", 175.50),
    (4, "Taylor Reed", "Culver City", 420.00),
    (5, "Chris Morgan", "Long Beach", 310.40),
    (6, "Avery Parker", "Pasadena", 198.95),
    (7, "Riley Scott", "Torrance", 515.30),
    (8, "Cameron Davis", "Burbank", 267.80)
]

con.executemany(
    "INSERT INTO customers (id, name, city, total_spent) VALUES (?, ?, ?, ?)",
    customers
)

con.commit()
print("8 fictional customer rows inserted and committed.")
```

```python
query = "SELECT * FROM customers"
df = pd.read_sql_query(query, con)
df
```

## MySQL-to-SQLite Difference

One difference I handled is data types. MySQL often uses types such as `VARCHAR` and `DECIMAL`, but for this SQLite version I used `TEXT` for text values and `REAL` for the numeric spending column. I also used `con.commit()` after the inserts so the new rows are saved.
