## Pounds to Kilograms Converter
This notebook demonstrates a Python function in Google Colab. The conversion rate is **1 pound = 0.453592 kilograms**.

**Before submitting:** Replace one example weight with your own weight in pounds. Run **Runtime → Run all** in Google Colab.

```python
def pounds_to_kilograms(pounds):
    return pounds * 0.453592

print("100 pounds =", pounds_to_kilograms(100), "kg")
print("156 pounds =", pounds_to_kilograms(156), "kg")
print("200 pounds =", pounds_to_kilograms(200), "kg")
```

### Explanation
The function accepts a weight in pounds, multiplies it by 0.453592, and returns the weight in kilograms. The three calls show that the same function works for different inputs.
