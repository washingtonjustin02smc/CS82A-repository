```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_clean.csv")
print(df.shape)
```

```python
print("Price mean:", df["price"].mean())
print("Price median:", df["price"].median())
print("Price standard deviation:", df["price"].std())
print("Qty mean:", df["qty"].mean())
print("Qty median:", df["qty"].median())
print("Qty standard deviation:", df["qty"].std())
```

The median value will be reported since the prices of the data are right-skewed. Some of the higher prices make the mean price equal to $55.52, making the median a regular price.

```python
df["price"].plot(kind="hist", bins=20)
plt.title("Distribution of Price")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

The price data is right-skewed.

```python
df["revenue"] = df["price"] * df["qty"]
df.head()
```

```python
corr = df.corr(numeric_only=True)
print(corr)
```

The strongest off-diagonal correlation is price vs. revenue at about 0.82.

```python
df.plot.scatter(x="qty", y="revenue")
plt.title("Quantity vs. Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")
plt.show()
```

There is a close connection between price and revenue. The correlation between them is around 0.82, which means that an increase in price leads to an increase in revenue. However, quantity sold and expensive products may affect this correlation too.
