Module 7 Lab 2 — Problem 7.1: Restaurant Ratings Regression

Dataset: `ratings.tsv` | Repo: https://github.com/washingtonjustin02smc/CS82A-repository/tree/module7

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score

df = pd.read_csv("ratings.tsv", sep="\t")
print("Restaurant records:", len(df))
print(df.to_string(index=False))
```

```python
features = ["ambience", "food", "service"]
X = df[features]
y = df["rating"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)
model = LinearRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Training records:", len(X_train))
print("Test records:", len(X_test))
print("Intercept:", round(model.intercept_, 4))
print("Test MAE:", round(mean_absolute_error(y_test, y_pred), 4))
print("Test R²:", round(r2_score(y_test, y_pred), 4))
```

```python
ranking = pd.DataFrame({
    "Factor": features, "Coefficient": model.coef_
}).sort_values("Coefficient", ascending=False).reset_index(drop=True)
ranking.index += 1
ranking.index.name = "Rank"
print("Factors ranked by regression coefficient:")
print(ranking.to_string(float_format=lambda v: f"{v:.4f}"))
print("\nOne-point increase in a factor predicts its coefficient's change in")
print("overall rating, holding the other two factors constant.")
```

```python
top = ranking.iloc[0]
runner = ranking.iloc[1]
last = ranking.iloc[2]
recommendation = (
    f"I would work on {top['Factor']} first because its coefficient "
    f"({top['Coefficient']:.4f}) is the largest, compared with "
    f"{runner['Factor']} ({runner['Coefficient']:.4f}) and "
    f"{last['Factor']} ({last['Coefficient']:.4f}). "
    f"This means a one-point increase in {top['Factor']} is associated with "
    f"the biggest predicted improvement in overall rating when the other scores "
    f"stay the same. Since these are observational ratings from only "
    f"{len(df)} restaurants, I would treat that as a starting priority rather "
    f"than proof that changing it will cause ratings to increase."
)
print("Recommendation:")
print(recommendation)
```
