```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("custdata.tsv", sep="\t")

excluded = df["housing.type"].isna().sum()
housing = df["housing.type"].dropna()
housing_counts = housing.value_counts().sort_values(ascending=False)

print(housing_counts)
print(f"Rows excluded: {excluded}")
```

56 rows were excluded because the housing type was missing.

```python
housing_counts.plot(kind="bar", figsize=(10, 6))
plt.title("Most Customers Own With a Mortgage/Loan or Rent")
plt.xlabel("Housing Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```
