```python
import pandas as pd

df = pd.read_csv("allbacks.csv")
df
```

```python
ordered_covers = sorted(df["cover"].tolist())
median_cover = ordered_covers[len(ordered_covers) // 2]
median_cover
```

**Median cover type:** The median cover type is paperback which means the middle cover type in the ordered data is paperback.

```python
mean_weight = df["weight"].mean()
mean_weight
```

**Mean weight:** The average weight of all the books is **706.67 grams**, which represents the typical weight of a book in this dataset.

```python
mean_by_cover = df.groupby("cover")["weight"].mean()
mean_by_cover
```

**Hardback vs. paperback:** Hardback books have an average weight of **796.43 grams**, paperback books average **628.13 grams**. That means that hardback books tend to weigh more than paperback books.
