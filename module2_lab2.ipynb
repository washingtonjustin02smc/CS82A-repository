**Hands-On Problem 4.2: Multiplication the long way**

This notebook demonstrates multiplication using repeated addition. It uses a `for` loop and a `while` loop without using the multiplication operator.

## Problem

Write functions that multiply two whole numbers without using the `*` operator. Both functions repeatedly add one number to itself. One function uses a `for` loop and the other uses a `while` loop.

```python
def multiply_for(number, times):
    result = 0
    for _ in range(times):
        result += number
    return result

print("For loop result:", multiply_for(7, 6))
```

```python
def multiply_while(number, times):
    result = 0
    count = 0
    while count < times:
        result += number
        count += 1
    return result

print("While loop result:", multiply_while(7, 6))
```

## Testing Both Functions

The two functions should agree with each other and with the expected product.

```python
number = 7
times = 6

for_result = multiply_for(number, times)
while_result = multiply_while(number, times)

print("7 + 7 + 7 + 7 + 7 + 7 =", for_result)
print("For loop result:", for_result)
print("While loop result:", while_result)
print("Both functions agree:", for_result == while_result)
```

```python
number = 5
times = 8

for_result = multiply_for(number, times)
while_result = multiply_while(number, times)

print("5 + 5 + 5 + 5 + 5 + 5 + 5 + 5 =", for_result)
print("For loop result:", for_result)
print("While loop result:", while_result)
print("Both functions agree:", for_result == while_result)
```

## How the While Loop Stops

The while loop uses a counter that starts at 0. Each time through the loop, the counter increases by 1. When the counter reaches the number of repetitions, the condition becomes false and the loop stops.
