# Module 2 Lab 2 - Multiplication the Long Way

**Hands-On Problem 4.2**

The textbook asks for a multiplication loop to be placed inside a function, called with two number arguments, and used to display the multiplication result.

For this lab, I show the idea twice:
- once with a `for` loop
- once with a `while` loop

Both functions multiply using repeated addition instead of the multiplication operator.

**Repository:** https://github.com/washingtonjustin02smc/CS82A-repository

## 1. Multiplication with a `for` loop

The result starts at 0. The loop adds the first number to the result once for each repetition given by the second number.

```python
def multiply_for(number, times):
    result = 0
    for count in range(times):
        result += number
    return result

for_answer = multiply_for(7, 6)
print("For loop result:", for_answer)
```

## 2. Multiplication with a `while` loop

The `while` version uses a counter. The counter starts at 0 and increases by 1 each time through the loop. The loop stops when the counter is no longer less than the second argument.

```python
def multiply_while(number, times):
    result = 0
    count = 0

    while count < times:
        result += number
        count += 1

    return result

while_answer = multiply_while(7, 6)
print("While loop result:", while_answer)
```

## 3. Test both functions

For the required example, both functions should produce 42 for 7 by 6.

```python
expected_answer = 42

print("Expected result:", expected_answer)
print("For loop result:", multiply_for(7, 6))
print("While loop result:", multiply_while(7, 6))
print("For loop is correct:", multiply_for(7, 6) == expected_answer)
print("While loop is correct:", multiply_while(7, 6) == expected_answer)
print("Both functions agree:", multiply_for(7, 6) == multiply_while(7, 6))
```

## 4. Additional test

This second test checks that both functions also work with another pair of whole numbers.

```python
expected_answer = 40

print("Expected result:", expected_answer)
print("For loop result:", multiply_for(5, 8))
print("While loop result:", multiply_while(5, 8))
print("Both functions agree:", multiply_for(5, 8) == multiply_while(5, 8))
```

## How the `while` loop stops

The `while` loop stops because `count` increases by 1 during every repetition. Once `count` reaches `times`, the condition `count < times` becomes false, so the loop ends.
