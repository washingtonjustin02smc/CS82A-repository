```python
import pandas as pd

df = pd.read_csv("messy_sales.csv")
print(df.shape)
print(df.head(10).to_string(index=False))
print(df.dtypes)
```

1. Certain values are not there in the price field.

2. The ZIP codes have been entered in numeric form which made certain leading zeroes disappearing.

3. The dates have been entered in varying formats and will have to be converted to one format.

```python
print(df.isna().sum())
```

```python
fill_value = df["price"].median()
df["price"] = df["price"].fillna(fill_value)
print(fill_value)
print(df.isna().sum())
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
print(df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
print(df["zip"].head(10).to_string(index=False))
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
print(df.dtypes)
```

```python
print(df.shape)
print(df["zip"].str.len().eq(5).all())
print(df["date"].dtype)
```

```python
return_mask = df["qty"] < 0
print(df.loc[return_mask].to_string(index=False))
print("Negative quantities retained:", return_mask.sum())
```

There were three entries where the values were in negative figures. I opted to retain such entries as they could be representing customer’s refunds rather than errors in the data. Changing them may lead to incorrect sales data.

```python
df.to_csv("sales_clean.csv", index=False, date_format="%Y-%m-%d")
print(df.shape)
print("sales_clean.csv saved")
```

## Cleaning Log

The database contained 300 rows. Missing prices were found in 12 rows, but I decided to use the median of $37.53 to fill the blanks since the median price is more robust compared to outliers. I found and deleted 8 duplicated rows, and now there are 292 rows remaining.

I changed ZIP codes to string data type and increased its length up to five digits to keep leading zeros. The date was converted to date time data type.

Three rows contained negative quantities, but I did not delete these rows because they might show returned products, rather than errors.

Bad choice in cleaning process will make a company receive inaccurate results from sales.
