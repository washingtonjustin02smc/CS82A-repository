```python
import pandas as pd

df = pd.read_csv('custdata.tsv', sep='\t')
df.head()
```

```python
columns = ['age', 'income', 'num.vehicles']
before_corr = df[columns].corr()
before_corr
```

```python
invalid_age = df['age'] > 100
negative_income = df['income'] < 0

print('Invalid ages over 100:', invalid_age.sum())
print('Negative incomes:', negative_income.sum())
print('Rows removed:', (invalid_age | negative_income).sum())
```

```python
clean = df[(df['age'] <= 100) & (df['income'] >= 0)].copy()
after_corr = clean[columns].corr()
after_corr
```

**What I removed and why:**  
I removed 8 rows where the age was over 100 because those ages were not realistic for this customer dataset. I also removed 1 row with a negative income because income should not be negative. These values could give wrong results.

**Interpretation:**  
After cleaning the data, the connection between age and income was weak. The relationship between age and number of vehicles was also weak. Income and number of vehicles had a positive correlation, meaning customers with higher incomes tended to have more vehicles, but the relationship wasn’t strong.
